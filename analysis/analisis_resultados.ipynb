```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Cargar datos

datos = pd.read_csv("../analysis/preentrega4_datos_recall.csv", sep = ",")
```

```python
# Primera columna → eje X
x = datos.iloc[:, 0]

plt.figure(figsize=(8, 4.5))

# Todas las columnas restantes → líneas
for columna in datos.columns[1:]:
    plt.plot(
        x,
        datos[columna],
        marker="o",
        label=columna
    )

plt.xlabel(datos.columns[0])
plt.ylabel("Recall@5 (%)")
plt.ylim(50, 105)
plt.grid()
plt.legend()
plt.tight_layout()

plt.savefig("recall_at_5.png", dpi=300, bbox_inches="tight")

plt.show()
```
